**Triple quotes**
- In jupiter notebook we have a **markdown** option to write information.
- **markdown** means we are conveying some text message.
- But in other platforms like **VS code, Pycharm** we don't have markdown option. so, we use triple quotes to convey information.
- This is also called **Doc string**.
- For the code part we use **single** and **double quotes** but for conveying the information we use **triple quotes**.
- **Triple quotes** are using in prompt engineering.
