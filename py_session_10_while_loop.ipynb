- Any loop we required 3 things
    - Initial point (start point)
    - increment or decrement
    - condition to stop
- in for loop the above 3 are happens in a single line
- but in while loop all 3 needs 3 different lines

```python
# start point
# while <condition>:
#     <start the code>
#      increment or decrement
range(1,10,1)
# start=1   stop=10-1=9   increment=1
# syntax
i=1
while <condition>:
    i=i+1
```

```python
i=1
while i<10:
    print(i)
    i=i+1
```

```python
i=1               # start
while i<10:       # condition i<10 which is True and loop starts
    print(i) 
    i=i+2         # increment 
```

```python
i=0
while i<11:
    print(i)
    i=i+2
```

```python
i=1
while i<11:
    print(i,end=' ')
    i+=1
```

```python
import random
i=0
while i<3:
    num=random.randint(1,10)
    if num==0:
        print(f'i{num} is zero')
    elif num%2==0:
        print(f'{num} is even')
    else:
        print(f'{num} is odd')
    i+=1
```
