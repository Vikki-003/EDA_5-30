**module**
- module is a single python file having variables, functions, classes, methods.

**package**
- package contains multiple python files 
- assume that ramesh has created 4 python files 
   - add.py
   - sub.py
   - mul.py
   - div.py
- All these 4 python files can save into a package

```python
# the package flow is 
math_function/
  add.py
  sub.py
  mul.py
  div.py

# Here we are importing the add, sub, mul, div from the package math_function so "from" refers from where the python file is going to import from.
from math_function import add
from math_function import sub
from math_function import mul
from math_function import div
```
