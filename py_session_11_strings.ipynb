**in vs range**

- if we want to access the elements directly, use **in** operator
     - for i in 'python' where i is character ex: p, y, t....
- if you want to assess the element by range, use **index**
     - for i in range(len(python)) where i is number which is used as index ex: s1[0],s1[1]....

```python
s1='python'
for i in s1:
    print(i,end=' ')
```

```python
s1='python'
for i in range(len(s1)):
    print(s1[i],end=' ')
```

```python
s='ola ola ola ola'
c=0
for i in range(len(s)):
    if s[i]=='a':
        c+=1
print(f'the a are {c}')
```

```python
s='ola ola ola ola'
c=0
summ=0
for i in range(len(s)):
    if s[i]=='a':
        c+=1
        summ+=i                  # here in summ we adding the index of a i represent the a's index by using if condithin
print(f'the a are {c}')
print(f'the sum of all a index is {summ}')
```
