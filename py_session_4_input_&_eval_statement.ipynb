**Round**
- round off it takes the nearest value when we use **round()**

```python
round(10.6789)  # the nearest value of 10.6789 is 11
```

**round(10.6789,n)**
- n indicates that after decimal n digits will print from the number that too round off.

```python
round(10.6789,1)  # this means  the 1 indicates after decimal only one digit from 10.6789.
```

```python
round(10.6789,2)  # 2 indicates after decimal only two digits from 10.6789
```

```python
a=10
b=21
c=36
avg=(a+b+c)/3
rounded_avg=round(avg,2)
print(f'the rounded average of {a},{b},{c} is: {rounded_avg}')
```
