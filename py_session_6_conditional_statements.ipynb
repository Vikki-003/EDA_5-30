**if-elif-else**

- When there is one condithon is there
     - We expect two answers
          - first answer under if block
          - second answer under else block
  
- When there are two conditions
     - We can expect three answers
          - first answer under if block based on condition-1
          - second answer under elif block based on condition-2
          - third answer under else block based on condition-3

```python
# syntax
if <condition>:
    <code>
elif <condition>:
    <code>
else:
    <code>
```

```python
num=eval(input('Enter number:'))
if num==1:
    print('one')
elif num==2:
    print('two')
elif num==3:
    print('three')
else:
    print('Enter a valid number')
```

```python
percentage=eval(input('Enter percentage:'))
if percentage>75:
    print('A grade')
elif 75<= percentage >60:
    print('B grade')
elif 60<= percentage >45:
    print('C grade')
elif 45<= percentage >30:
    print('D grade')
else:
    print('fail')
```
