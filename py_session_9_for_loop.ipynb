**loops**

- loops are used to execute a block of code multiple times
- we have to think  about 3 parameters in loop
   - start point
   - increment or decriment
   - condition to stop

- we have 2 loops 
- for loop and while loop
- if we know the condition when to stop the loop then **for loop** is prefered
- if we dont know when to stop the loop then **while loop** is prefered

**case-1: range(stop)**

- atarting point is default by 0
- direction: positive direction
- positive direction means **increment**
- by default increment by **one**
- last point is stop-1

**case-3: range(start,stop,step)**

- start value is = start
- if step is +ve sign
   - it is positive direction
   - increment by **step value**
   - last value is = stop-1
- if step is -ve sign
   - it is negative direction(decrement)
   - decrement by **step value**
   - last value is = stop+1

```python
for i in range(1,20,2):
    print(i,end=' ')
```

```python
for i in range(20,1,-2):
    print(i,end=' ')
```

```python
for i in range(-20,-1,2):
    print(i,end=' ')
```

```python
for i in range(20,1,2):
    print(i,end=' ')
```

```python
import random
for i in range(3):
    num1=random.randint(1,10)
    num2=eval(input('enter the number:'))
    if num1==num2:
        print('you won')
        break
    else:
        print('you lost')
```

```python
# case-1: we repeated 3 times irrespective of win or loss
# case-2: we applied a break when a successful attempt
# case-3: every time you lose 
#           it should print how many chances are left
# case-4: if you lose all chances it should print try again after 24hrs

# case-3
chance=3
for i in range(3):
    num1=random.randint(1,10)
    num2=eval(input('enter the number:'))
    if num1==num2:
        print('you won')
        break
    else:
        print('you lost')
        chance= chance-1
        print(f'you have left {chance} chances')
```

```python
# case-4
chance=3
for i in range(3):
    num1=random.randint(1,10)
    num2=eval(input('enter the number:'))
    if num1==num2:
        print('you won')
        break
    else:
        print('you lost')
        chance= chance-1
        print(f'you have left {chance} chances')
        if chance==0:
            print('try after 24hrs')
```

**continue-pass-break**

```python
for i in range(5):
    if i==2:
        print('great')
    print(i)

# step-1 ===> i=0   0==2 (no)  it prints ==> 0
# step-2 ===> i=1   1==2 (no)  it prints ==> 1
# step-3 ===> i=2   2==2 (yes)  it prints 'great' and loop exit out side i should print so print ==> 2
# step-4 ===> i=3   3==2 (no)  it prints ==> 3
# step-5 ===> i=4   4==2 (no)  it prints ==> 4
```

```python
for i in range(5):
    if i==2:
        print('great')
        continue
    print(i) 
```

- when continue is using when it comes it will stop and start again the next loop. it will leave the remaing part
